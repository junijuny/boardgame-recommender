```python
import pandas as pd

df = pd.read_csv(r'..\data\clean_games.csv')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=1000, stop_words="english")
tfidf_matrix = tfidf.fit_transform(df["Description"])

print(tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# 예: 'Catan' 게임의 인덱스 찾기
target_game_name = "Catan"
target_idx = df[df["Name"].str.lower() == target_game_name.lower()].index[0]

# 기준 게임 벡터와 전체 게임 간 유사도 계산
target_vector = tfidf_matrix[target_idx]
similarity_scores = cosine_similarity(target_vector, tfidf_matrix).flatten()
```

```python
import numpy as np

# 유사도 순으로 인덱스 정렬 (내림차순)
similar_indices = np.argsort(similarity_scores)[::-1]

# 자기 자신(0번째)을 제외한 상위 5개 추출
top5_indices = similar_indices[1:6]

for rank, idx in enumerate(top5_indices, 1):
    game_name = df.iloc[idx]["Name"]
    score = similarity_scores[idx]
    print(f"{rank}위: {game_name} (유사도: {score:.4f})")
```
