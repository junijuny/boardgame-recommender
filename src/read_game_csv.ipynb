```python
# column 선별
selected_cols = [
    'BGGId', 'Name', 'YearPublished', 'ImagePath',
    'GameWeight', 'Description',
    'MinPlayers', 'MaxPlayers', 'BestPlayers', 'ComMinPlaytime', 'ComMaxPlaytime',
    'BayesAvgRating', 'NumUserRatings'
]
game_df = game_df[selected_cols].copy()
```

```python
# name, Dedctiption 결측치 확인 및 결측행 삭제
game_df[['Name', 'Description']].isna().sum()
game_df = game_df.dropna(subset=['Name', 'Description'])
empty_desc = game_df['Description'].str.strip() == ''
game_df = game_df[~empty_desc]
```

```python
# index 제정렬
game_df.reset_index(drop=True, inplace=True)
```

```python
# 플레이타임 필터링 & 인원수 이상치 확인
condition = (
    (game_df['ComMinPlaytime'] > 0) & 
    (game_df['ComMaxPlaytime'] < 1000) &
    (game_df['MinPlayers'] >= 1) & 
    (game_df['MinPlayers'] <= 10)
)
game_df = game_df[condition]
```

```python
# BestPlayers 값 보정
mask = game_df['BestPlayers'] == 0
game_df.loc[mask, 'BestPlayers'] = (game_df.loc[mask, 'MinPlayers'] + game_df.loc[mask, 'MaxPlayers']) // 2
```

```python
# 수정된 data 확인
game_df.info()

# 결측치 확인
print("남은 결측치 수:\n", game_df.isna().sum())

# 수치형 data 통계 확인
game_df[['MinPlayers', 'MaxPlayers', 'BestPlayers', 'ComMinPlaytime', 'ComMaxPlaytime']].describe()
```
